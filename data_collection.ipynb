The criteria for ticker seleccion correspond to stocks with a Price(day) below $5, a Volume more than 100k, and New York Stock Exchange (NYSE) and NASDAQ as Exchange.

```python
#Combine all tickers from each page
all_tickers = [page1, page2, page3, page4]
```

```python
# remove duplicate tickers
tickers = list(dict.fromkeys(
    ticker.strip().upper()
    for page in all_tickers
    for ticker in page
    if ticker.strip()
))
print('Total unique tickers: ', len(tickers))
```

```python
start_date = "2020-01-01"
end_date = "2026-01-01"
```

```python
base_dir = Path('penny_stocks_research')
raw_dir = base_dir / 'raw'
raw_dir.mkdir(parents=True, exist_ok=True)
print('Saving files in:', base_dir.resolve())
```

```python
#testing yf for data collection
test = yf.Ticker("SNDL").history(
    start=start_date,
    end=end_date,
    interval="1d",
    auto_adjust=False,
    actions=True
)

display(test.head(5))

print("Number of rows:", len(test))
print("Columns:", test.columns.tolist())
```

```python
# Downloading tickers
all_stocks=[]
download_log=[]

for i, ticker in enumerate(tickers , start=1):
    print(f"{i}/308: {ticker}")
    try:
        history= yf.Ticker(ticker).history(
            start=start_date,
            end=end_date,
            interval='1d',
            auto_adjust=False,
            actions=True,
            repair=True
        )
        if history.empty:
            download_log.append({
                'ticker': ticker ,
                'statues': 'EMPTY',
                'rows':0
            })
            continue
        history.index = pd.to_datetime(history.index)
        if history.index.tz is not None:
            history.index = history.index.tz_localize(None)
        history.index.name = 'date'
        df = history.reset_index()
        df['ticker'] = ticker
        df.to_csv(raw_dir / f'{ticker}.csv', index=False)

        all_stocks.append(df)
        download_log.append({
            'ticker':ticker,
            'status':'OK',
            'rows': len(df)
        })
    except Exception as e:
        print('Error:', ticker,str(e))
        download_log.append({
            'ticker':ticker,
            'status':'Failed',
            'rows':0
        })
log_df = pd.DataFrame(download_log)
display(log_df)
log_df.to_csv(base_dir / 'download_log.csv', index=True)

if all_stocks:
    panel = pd.concat(all_stocks, ignore_index=True)
    panel = panel.sort_values(['ticker','date'])
    panel.to_csv( base_dir / 'historical_stock_data.csv', index=False)
    print('Total observations:', len(panel))
    print('Succefully downloaded', len(all_stocks))
    display(panel.head())
else:
    print('No data downloaded. Check for errors in the code')
```

```python
# Use your existing combined DataFrame
panel = panel.copy()
panel["date"] = pd.to_datetime(panel["date"])

coverage = (
    panel.groupby("ticker")
    .agg(
        observations=("date", "nunique"),
        first_date=("date", "min"),
        last_date=("date", "max"),
        median_close=("Close", "median"),
        median_volume=("Volume", "median")
    )
    .reset_index()
    .sort_values("observations")
)

print("Total tickers:", len(coverage))
print("\nObservation-count summary:")
display(coverage["observations"].describe())

print("\nCoverage by threshold:")
for threshold in [60, 252, 500, 1000, 1250]:
    count = (coverage["observations"] >= threshold).sum()
    print(f"{threshold:4d}+ observations: {count} tickers")
```

```python
# keeping stocks with more than 252 observations(1 bussines year)
eligible_tickers = coverage.loc[coverage['observations'] >=252, 'ticker'].tolist()
```

```python
#filter tickers
panel_filtered = panel[panel['ticker'].isin(eligible_tickers)]
# save them into a csv file
panel_filtered.to_csv('penny_stocks_research/historical_stock_data_252days.csv', index=False)
```

```python
panel_filtered.head(5)
```

```python
price_summary = (
    panel_filtered.groupby("ticker")
    .agg(
        observations=("Close", "count"),
        pct_days_below_5=(
            "Close",
            lambda s: 100 * (s <= 5).mean()
        ),
        median_close=("Close", "median"),
        median_volume=("Volume", "median")
    )
    .reset_index()
)

# Review stocks that stayed at or below $5 on most observed days
display(
    price_summary.sort_values(
        "pct_days_below_5",
        ascending=False
    ).head(30)
)
```

Observations:

- Historical coverage: Most displayed stocks have 1,508 observations.

- Historical price eligibility: several stocks were below or equal to $5 on 100% of their observed days.

- Liquidity varies substantially: Low volume can make price changes noisier, and trading strategies harder to execute. As observed, ABEV has median daily volume of about 22.6 million shares, while SEATW has only 1,700. 


```python
# saving price summary
price_summary.to_csv(
    "historical_price_summary.csv",
    index=False
)
```
